```python
from pysplines import BSplineCurve, plot_curve_2d, plot_basis

import numpy as np
import matplotlib.pyplot as plt
```

## Partition of Unity

The partition of unity is one of the fundamental properties of B-Splines:

$$
\sum_{i=1}^{n}{N_{i,p}(\xi) = 1}
$$

Since the local influence of a basis $N_{i,p}$ extends only in the interval $[\xi_{i}, \xi_{i+p+1}]$, an open (clamped) knot vector is required for the partition of unity to extend to the boundaries.

```python
points = np.array([
        [1.0, 0.5], 
        [2.0, 2.0], 
        [4.0, 2.0],
        [3.0, 1.0], 
        [5.0, 0.0]
    ])
```

In this example, only $N_1$ is acting on the first knot interval, and only $N_1$ and $N_2$ on the second. Therefore, this is insufficient to fulfill the partition of unity. We would have the origin interpolated if we considered this part of the domain!

```python
import numpy as np

p = 2
knots = [0, 1, 2, 3, 4, 5, 6, 7]

spline = BSplineCurve(knots, p)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Plot basis functions on left
plot_basis(spline, ax=ax1)
plot_curve_2d(spline, points, ax=ax2)

plt.tight_layout()
plt.show()
```

```python
p = 2
knots = [0, 0, 0, 3, 3, 7, 7, 7]

spline = BSplineCurve(knots, p)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Plot basis functions on left
plot_basis(spline, ax=ax1)

# Plot curve on right
plot_curve_2d(spline, points, ax=ax2)

plt.tight_layout()
plt.show()
```

```python
p = 2
knots = [0, 0, 0, 3, 3, 3, 7, 7, 7]
points = np.append(points, [[6.0, 1.5]], axis=0)

spline = BSplineCurve(knots, p)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Plot basis functions on left
plot_basis(spline, ax=ax1)

# Plot curve on right
plot_curve_2d(spline, points, ax=ax2)

plt.tight_layout()
plt.show()
```
